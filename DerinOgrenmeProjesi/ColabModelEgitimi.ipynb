```python
import os
import cv2
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers, regularizers
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
# --- 1. Veri Yolları ve Parametreler ---
DATA_DIR = "/content/dataset/"
IMAGE_DIR = os.path.join(DATA_DIR, 'images', '0a_image_irfs')
MASK_DIR = os.path.join(DATA_DIR, 'masks', '1a_label_irfs')

IMG_HEIGHT = 256
IMG_WIDTH = 256
NUM_CLASSES = 6
BATCH_SIZE = 16
EPOCHS = 200

# --- 2. Veri Yükleme ve Ön İşleme ---
def load_and_preprocess_data(image_dir, mask_dir, img_height, img_width, num_classes):
    images = []
    masks = []

    if not os.path.exists(image_dir) or not os.path.exists(mask_dir):
        print("Hata: Veri yolu bulunamadı.")
        return np.array(images), np.array(masks)

    image_files = sorted(os.listdir(image_dir))
    print(f"{len(image_files)} görüntü bulundu.")

    value_map = {np.uint8(0): 0, np.uint8(14): 1, np.uint8(38): 2, np.uint8(52): 3, np.uint8(75): 4, np.uint8(113): 5}

    for img_file in image_files:
        img_path = os.path.join(image_dir, img_file)
        img = cv2.imread(img_path)
        if img is None: continue
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = cv2.resize(img, (img_width, img_height)) / 255.0
        images.append(img)

        mask_file = os.path.splitext(img_file)[0] + '.png'
        mask_path = os.path.join(mask_dir, mask_file)
        if not os.path.exists(mask_path): continue

        mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)
        if mask is None: continue
        mask = cv2.resize(mask, (img_width, img_height), interpolation=cv2.INTER_NEAREST)
        mapped_mask = np.vectorize(value_map.get)(mask)
        mapped_mask_one_hot = to_categorical(mapped_mask, num_classes=num_classes)
        masks.append(mapped_mask_one_hot)

    return np.array(images), np.array(masks)

# --- 3. U-Net Modeli ---
def build_light_unet(input_shape, num_classes):
    def conv_block(x, filters):
        x = layers.Conv2D(filters, 3, activation='relu', padding='same',
                          kernel_regularizer=regularizers.l2(0.001))(x)
        x = layers.Dropout(0.3)(x) # Dropout eklendi
        x = layers.Conv2D(filters, 3, activation='relu', padding='same',
                          kernel_regularizer=regularizers.l2(0.001))(x)
        return x

    inputs = layers.Input(input_shape)

    # Encoder
    c1 = conv_block(inputs, 32)
    p1 = layers.MaxPooling2D((2, 2))(c1)

    c2 = conv_block(p1, 64)
    p2 = layers.MaxPooling2D((2, 2))(c2)

    c3 = conv_block(p2, 128)
    p3 = layers.MaxPooling2D((2, 2))(c3)

    c4 = conv_block(p3, 256)
    p4 = layers.MaxPooling2D((2, 2))(c4)

    c5 = conv_block(p4, 512)
    c5 = layers.Dropout(0.4)(c5) # Bottleneck'e Dropout eklendi

    # Decoder
    up6 = layers.concatenate([layers.Conv2DTranspose(256, 2, strides=2, padding='same')(c5), c4])
    c6 = conv_block(up6, 256)

    up7 = layers.concatenate([layers.Conv2DTranspose(128, 2, strides=2, padding='same')(c6), c3])
    c7 = conv_block(up7, 128)

    up8 = layers.concatenate([layers.Conv2DTranspose(64, 2, strides=2, padding='same')(c7), c2])
    c8 = conv_block(up8, 64)

    up9 = layers.concatenate([layers.Conv2DTranspose(32, 2, strides=2, padding='same')(c8), c1])
    c9 = conv_block(up9, 32)

    outputs = layers.Conv2D(num_classes, 1, activation='softmax')(c9)

    return models.Model(inputs, outputs)

# --- 4. Eğitim ve Görselleştirme ---
if __name__ == '__main__':
    print("Veriler yükleniyor...")

    images, masks = load_and_preprocess_data(IMAGE_DIR, MASK_DIR, IMG_HEIGHT, IMG_WIDTH, NUM_CLASSES)

    if len(images) == 0 or len(masks) == 0:
        print("Geçerli veri bulunamadı.")
        exit()

    X_train, X_test, y_train, y_test = train_test_split(images, masks, test_size=0.2, random_state=42)

    model = build_light_unet((IMG_HEIGHT, IMG_WIDTH, 3), NUM_CLASSES)
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    model.summary()

    early_stopping = EarlyStopping(
        monitor='val_loss',
        patience=20,
        restore_best_weights=True,
        verbose=1
    )

    checkpoint_filepath = '/content/drive/MyDrive/colab_models/unet_segmentation_best_model.h5' # Kendi yolunuzu güncelleyin
    model_checkpoint_callback = tf.keras.callbacks.ModelCheckpoint(
        filepath=checkpoint_filepath,
        save_weights_only=False, # Tüm modeli kaydet
        monitor='val_loss',      # val_loss'u izle
        mode='min',              # En düşük val_loss'u kaydet
        save_best_only=True,     # Sadece en iyi modeli kaydet
        verbose=1
    )

    print("Veri Artırma jeneratörleri oluşturuluyor...")
    from tensorflow.keras.preprocessing.image import ImageDataGenerator

    data_gen_args = dict(
        rotation_range=15,
        width_shift_range=0.1,
        height_shift_range=0.1,
        shear_range=0.1,
        zoom_range=0.1,
        horizontal_flip=True,
        vertical_flip=False,
        fill_mode='nearest'
    )

    image_datagen = ImageDataGenerator(**data_gen_args)
    mask_datagen = ImageDataGenerator(**data_gen_args)

    seed = 42 # Aynı rastgelelik için seed
    train_image_generator = image_datagen.flow(X_train, batch_size=BATCH_SIZE, seed=seed, shuffle=True)
    train_mask_generator = mask_datagen.flow(y_train, batch_size=BATCH_SIZE, seed=seed, shuffle=True)

    def create_combined_generator(img_gen, mask_gen):
        while True:
            img_batch = next(img_gen)
            mask_batch = next(mask_gen)
            yield img_batch, mask_batch

    combined_train_generator = create_combined_generator(train_image_generator, train_mask_generator)

    steps_per_epoch = len(X_train) // BATCH_SIZE
    if len(X_train) % BATCH_SIZE != 0:
        steps_per_epoch += 1

    print("Model eğitiliyor...")
    history = model.fit(
        combined_train_generator, # Artırılmış veri jeneratörünü kullan
        steps_per_epoch=steps_per_epoch,
        epochs=EPOCHS,
        validation_data=(X_test, y_test),
        callbacks=[early_stopping, model_checkpoint_callback], # ModelCheckpoint'i de ekledik
        verbose=1
    )

    print("Model eğitimi tamamlandı.")

    # Performans Grafikleri
    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Eğitim Doğruluğu')
    plt.plot(history.history['val_accuracy'], label='Doğrulama Doğruluğu')
    plt.title('Model Doğruluğu')
    plt.xlabel('Epoch')
    plt.ylabel('Doğruluk')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Eğitim Kaybı')
    plt.plot(history.history['val_loss'], label='Doğrulama Kaybı')
    plt.title('Model Kaybı')
    plt.xlabel('Epoch')
    plt.ylabel('Kayıp')
    plt.legend()
    plt.show()

    # Modeli Kaydet
    model_save_path = 'unet_segmentation_model.h5'
    model.save(model_save_path)
    print(f"Model şuraya kaydedildi: {model_save_path}")

    # Tahmin Görselleştirme
    print("\nTest setinden örnek tahminler görselleştiriliyor...")
    num_display_samples = 3

    for i in range(num_display_samples):
        idx = np.random.randint(0, len(X_test))
        sample_image = X_test[idx]
        true_mask = y_test[idx]

        predicted_mask = model.predict(np.expand_dims(sample_image, axis=0))[0]
        predicted_mask_indices = np.argmax(predicted_mask, axis=-1)
        true_mask_indices = np.argmax(true_mask, axis=-1)

        plt.figure(figsize=(15, 5))
        plt.subplot(1, 3, 1)
        plt.imshow(sample_image)
        plt.title('Orijinal Görüntü')
        plt.axis('off')

        plt.subplot(1, 3, 2)
        plt.imshow(true_mask_indices, cmap='jet', vmin=0, vmax=NUM_CLASSES-1)
        plt.title('Gerçek Maske')
        plt.axis('off')

        plt.subplot(1, 3, 3)
        plt.imshow(predicted_mask_indices, cmap='jet', vmin=0, vmax=NUM_CLASSES-1)
        plt.title('Tahmin Edilen Maske')
        plt.axis('off')
        plt.show()
```
